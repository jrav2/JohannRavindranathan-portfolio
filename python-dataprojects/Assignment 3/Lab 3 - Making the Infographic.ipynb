Create Visualization

```python
import pandas as pd
import altair as alt

df = pd.read_csv("Pictures Spreadsheet 1-30.csv") #replace with your file name from above
```

```python
alt.Chart(df).mark_bar().encode(
    x=alt.X('date', title='Date (February 10 - February 14)'),
    y=alt.Y('percentage', title='Percentage Scale of Image', scale=alt.Scale(domain=[1,100])),
    color=alt.Color('colorhex', scale=None, legend=None),
    tooltip=['colorhex','percentage']
).configure_axis(
    grid=False
).properties(
    title='The Sky from the Quad, University of Washington'
)
```
