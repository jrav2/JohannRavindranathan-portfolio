```python
import glob

image_folder = '/content/drive/MyDrive/Geog 450 - Lab 3 Processed Photos/*.png'
image_files = glob.glob(image_folder)
```

```python
all_dfs = []

for file in image_files:

    im2 = Image.open(file)

    na = np.array(im2.convert('RGB'))

    colours, counts = np.unique(na.reshape(-1,3), axis=0, return_counts=1)

    myhex = []
    for i in range(colours.shape[0]):
        myhex.append(rgb_to_hex((colours[i,0], colours[i,1], colours[i,2])))

    totalpixel = counts.sum()

    df = pd.DataFrame({
        'colorhex': myhex,
        'numpixel': np.int_(counts),
        'percentage': (counts / totalpixel) * 100,
    })

    all_dfs.append(df)
```

```python
final_df = pd.concat(all_dfs, ignore_index=True)
final_df
```
